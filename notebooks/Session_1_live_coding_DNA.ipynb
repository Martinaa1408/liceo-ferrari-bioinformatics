## 1. DNA is a string

```python
dna = "ATGGTGCACCTGACTCCTGAGGAGAAG"   # start of the human HBB gene
print(len(dna))
print(dna.count("G"))
```

## 2. GC content

```python
def gc_content(seq):
    gc = seq.count("G") + seq.count("C")
    return gc / len(seq)

print(round(gc_content(dna), 2))
```

## 3. The genetic code as a lookup table

```python
bases = "TCAG"
amino = ("FFLLSSSSYY**CC*W"
         "LLLLPPPPHHQQRRRR"
         "IIIMTTTTNNKKSSRR"
         "VVVVAAAADDEEGGGG")
codons = [a + b + c for a in bases for b in bases for c in bases]
code = dict(zip(codons, amino))

print(len(code))
```

## 4. From DNA to protein

```python
def translate(seq):
    protein = ""
    for i in range(0, len(seq) - 2, 3):
        protein += code[seq[i:i + 3]]
    return protein

print(translate(dna))
```

## 5. One letter, one disease

```python
sickle = dna[:19] + "T" + dna[20:]   # A becomes T in codon 7

print(dna)
print(sickle)
print(translate(dna))
print(translate(sickle))
```
